```python
import pandas as pd
import numpy as np
import os

# Path to the raw data folder
RAW_DATA_PATH = "/content/drive/MyDrive/Meridian_Churn_Capstone/Data/Raw"

# File paths
customers_path = os.path.join(RAW_DATA_PATH, "customers.csv")
products_path = os.path.join(RAW_DATA_PATH, "products.csv")
orders_path = os.path.join(RAW_DATA_PATH, "orders.csv")
order_items_path = os.path.join(RAW_DATA_PATH, "order_items.csv")
support_tickets_path = os.path.join(RAW_DATA_PATH, "support_tickets.csv")

# Load the datasets
customers = pd.read_csv(customers_path)
products = pd.read_csv(products_path)
orders = pd.read_csv(orders_path)
order_items = pd.read_csv(order_items_path)
support_tickets = pd.read_csv(support_tickets_path)

print("All five datasets loaded successfully.")
```

```python
datasets = {
    "Customers": customers,
    "Products": products,
    "Orders": orders,
    "Order Items": order_items,
    "Support Tickets": support_tickets
}

for name, df in datasets.items():
    print(f"{name}: {df.shape[0]:,} rows × {df.shape[1]} columns")
```

```python
for name, df in datasets.items():
    print(f"\n{name}")
    print("-" * len(name))
    print(list(df.columns))
```

```python
def dataset_summary(df, dataset_name):
    """
    Display a simple summary of a dataset.
    """

    print(f"\nDataset: {dataset_name}")
    print(f"Rows: {df.shape[0]:,}")
    print(f"Columns: {df.shape[1]:,}")
    print(f"Missing values: {df.isna().sum().sum():,}")
    print(f"Duplicate rows: {df.duplicated().sum():,}")


# Use the function on all five datasets
for name, df in datasets.items():
    dataset_summary(df, name)
```

```python
for name, df in datasets.items():

    rows = df.shape[0]

    if rows >= 50000:
        size_category = "Very large dataset"
    elif rows >= 10000:
        size_category = "Large dataset"
    elif rows >= 1000:
        size_category = "Medium dataset"
    else:
        size_category = "Small dataset"

    print(f"{name}: {rows:,} rows → {size_category}")
```

```python
datasets_with_missing = [
    name for name, df in datasets.items()
    if df.isna().sum().sum() > 0
]

print("Datasets containing missing values:")
for name in datasets_with_missing:
    print("-", name)
```

```python
missing_values_summary = {
    name: df.isna().sum().sum()
    for name, df in datasets.items()
}

print("Missing values by dataset:")
for name, missing in missing_values_summary.items():
    print(f"{name}: {missing:,}")
```
