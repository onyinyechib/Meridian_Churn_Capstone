```python
print("1. Orders → Customers")
order_customer_check = orders["customer_id"].isin(customers["customer_id"])
print("All matched:", order_customer_check.all())
print("Unmatched:", (~order_customer_check).sum())

print("\n2. Order Items → Orders")
order_item_order_check = order_items["order_id"].isin(orders["order_id"])
print("All matched:", order_item_order_check.all())
print("Unmatched:", (~order_item_order_check).sum())

print("\n3. Order Items → Products")
order_item_product_check = order_items["product_id"].isin(products["product_id"])
print("All matched:", order_item_product_check.all())
print("Unmatched:", (~order_item_product_check).sum())

print("\n4. Support Tickets → Customers")
ticket_customer_check = support_tickets["customer_id"].isin(customers["customer_id"])
print("All matched:", ticket_customer_check.all())
print("Unmatched:", (~ticket_customer_check).sum())
```

```python
| Relationship                                            | Result         | Interpretation                                                                     |
| ------------------------------------------------------- | -------------- | ---------------------------------------------------------------------------------- |
| `Orders.customer_id` → `Customers.customer_id`          | ❌ 20 unmatched | 20 order records refer to customer IDs that are not currently found in `Customers` |
| `Order_Items.order_id` → `Orders.order_id`              | ✅ All matched  | Every order item refers to an existing order                                       |
| `Order_Items.product_id` → `Products.product_id`        | ✅ All matched  | Every order item refers to an existing product                                     |
| `Support_Tickets.customer_id` → `Customers.customer_id` | ✅ All matched  | Every support ticket refers to an existing customer                                |
```

```python
This gives us an important insight

The database structure is mostly connected correctly, but there is a referential-integrity problem between Orders and Customers.

In simple terms:

Customers
   │
   │ customer_id
   ▼
Orders

is supposed to work like this:

Every customer ID appearing in Orders should exist in Customers.
```

```python
unmatched_orders = orders[
    ~orders["customer_id"].isin(customers["customer_id"])
]

unmatched_orders
```

2.X Data Sources and Operational Systems

The five datasets provided for this project represent information that would plausibly be generated by different operational systems within Meridian Retail Group. Meridian operates as a multi-channel retailer, selling products through its website, mobile application, and physical stores. The datasets therefore represent different aspects of its customer, product, sales, and customer-support operations.

The Customers dataset (customers.csv) would plausibly originate from Meridian's Customer Relationship Management (CRM) or Customer Master System. This system would maintain customer profiles, including identification and contact details, location, membership tier, preferred communication channel, marketing consent, and referral source.

The Products dataset (products.csv) would plausibly come from a Product Information Management or Product Catalog System. Such a system would maintain information about the products Meridian sells, including product names, categories, sub-categories, brands, costs, selling prices, launch dates, and whether products are currently active.

The Orders dataset (orders.csv) would plausibly be generated by Meridian's Order Management System (OMS), which could consolidate transactions originating from its website, mobile application, and physical stores. It contains order-level information such as the customer who placed the order, order date, sales channel, order status, payment method, shipping cost, discount, and total order value.

The Order Items dataset (order_items.csv) would plausibly originate from the same transaction or order-management infrastructure. While the Orders table records information about the overall order, Order Items records the individual products included in each order, together with quantities and the prices paid at the time of purchase. This creates the link between individual orders and products.

Finally, the Support Tickets dataset (support_tickets.csv) would plausibly originate from Meridian's Customer Support or Help Desk System, potentially integrated with its CRM. It records customer-service interactions, including the issue category, communication channel, resolution time, satisfaction score, and description of the customer's issue.

These operational systems collectively provide the data required for the churn investigation. Customer information describes who the customers are, order and order-item data describe their purchasing behaviour, product data provides information about what they purchase, and support-ticket data provides information about their customer-service experience. Combining these sources therefore allows customer churn to be investigated from behavioural, transactional, product, and support perspectives.
